<h2>Convolutional layer</h2>

```python
import torch
from torch.nn.parameter import Parameter

A = torch.tensor([[[[1, 2, 0, 1],
                    [-1, 0, 3, 2],
                    [1, 3, 0, 1],
                    [2, -2, 1, 0]]]]).float()

conv2d = torch.nn.Conv2d(1, 1, kernel_size = 2, bias = False)
conv2d.weight = Parameter(torch.tensor([[[[1, -1], [-1, 1]]]]).float())

output = conv2d(A)

print(output)
```

<h2>Pooling</h2>

Pooling aims at reduce the computational complexity of the neural network

```python
import torch

A = torch.tensor([[
    [1, 2, -1, 1],
    [0, 1, -2, -1],
    [3, 0, 5, 0],
    [0, 1, 4, -3]
]]).float()

avg_pool = torch.nn.AvgPool2d(2)
out = avg_pool(A)

print(out.tolist())
```

<h2>Dropout</h2>

Regularization technique to overcome overfitting by *dropping random entries from the input tensor with probability* **p** during the training phase. 

The left tensor entries are multiplied by $\frac{1}{1-p}$.

In the evaluation phase the *droput layer does not affect the input tensor untouched*.

Dropout makes the network more robust because **it cannot rely on any particular set of input neurons for making predictions** $\rightarrow$ network tries to extract patterns from different parts of the dataset
